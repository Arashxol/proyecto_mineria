```python
# Convertimos variables categoricas a variables dummy
df = pd.get_dummies(df, columns=['Education','Marital_Status'])
#df['Education'] = df['Education']
print(df.shape)
df.head(2)
```

```python
# Convertimos a entero todos los campos que empiezan con 'Education' o 'Marital_Status'
for col in df.columns:
    if col.startswith('Education') or col.startswith('Marital_Status'):
        if df[col].dtype == bool:
            df[col] = df[col].astype(int)
df[[c for c in df.columns if c.startswith('Education') or c.startswith('Marital_Status')]].head(2)

print(df.shape)
df.head(2)
```

## 4.2. Modelo

```python
df.columns
```

```python
df.Response.unique()
```

```python
# seleccionamos variables explicativas y variables dependiente
X = df[['Kidhome','Recency']]
y = df['Response']
```

```python
# Dividir los datos en conjuntos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=777)
```

```python
# tamaño de datos de entrenamiento
print('tamaño train', X_train.shape)
print('tamaño de test', X_test.shape)
```

```python
# Ajusta el modelo de regresión logística
modelo_logit = sm.Logit(y_train, X_train).fit()
```

# 4.3. Evaluación

### 4.3.1. Resultados del Modelo

```python
# Muestra un resumen del modelo
print(modelo_logit.summary())
```

```python
# prediccion
y_pred_train = modelo_logit.predict(X_train)
y_pred_train.head(2)
```

```python
X_train.shape
```

```python
# Calcula la exactitud del modelo en el entrenamiento
# y_train: valores verdadaderos
# y_pred: valores predichos
exactitud_train = accuracy_score(y_train, (y_pred_train>0.5).astype(int))
print("exactitud del modelo:", exactitud_train)
```

# 4.3.2. Predicción

```python
# Realiza predicciones en el conjunto de datos de prueba
y_pred = modelo_logit.predict(X_test)
y_pred
```

```python
exactitud_test = accuracy_score(y_test, (y_pred>0.5).astype(int))
print("Exactitud del modelo:", exactitud_test)
```
